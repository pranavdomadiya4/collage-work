```python
s="B4A1D3"
s1=s2=output=""

for x in s:
    if x.isalpha():
        s1+=x
    else:
        s2+=x
for x in sorted(s1):
    output+=x
for x in sorted(s2):
    output+=x

output2="".join(sorted(s1)+sorted(s2))
print(output)
print(output2)
```
