## Tuple
-tupple is immutable. tuple is read only version of list
-if data is fixed and never changes then we can go for tuple
-insertion order is preserved
-duplicate are allowed
-heterogeneous abject are allowed
-we can differantiate duplicate object by using index
-tuple suppoerts both +ve and -ve indexing
-represent tuple elements within parenthesis(()) with comma seprator
-paranthesis are optional but recommended to use

```python
#creation of tuple

t=(10,)
s=(10)
print(type(t))
print(type(s))

t1=(1,2,3,4,6,8)
print(t1)
t2=tuple(range(10,20,2))
print(t2)
l=[1,2,3,5,8]
print(tuple(l))
```

```python
#accesing the element of tuple

#1.by using index
t=(10,20,30,40,50,60)
print(t[3])
# print(t[7])
print(t[-5])
print(t[-3])
#2.by using slicling operator
t1=(10,20,30,40,50,60)
print(t[2:5])
print(t[2:100])
print(t[:100])
print(t[::2])
print(t[4:])
print(t[:-3])
print(t[-2:])
print(t[-5:-2])
print(t[-5:-1:-1])
print(t[-5:-1:1])
print(t[5:1:-1])
```

```python
#mathmatical operation for tuple
# (+)-cancatenation
# (*)- repetation

t1=(10,20,30,40)
t2=(10,20,30,40)
print(t1+t2)
print(t1*3)
```

```python
#tuple packing
a=10
b=20
c=30
d=40
t=a,b,c,d
print(t)
```

```python
#unpacking
t=(10,20,30,40)
a,b,c,d=t
print(b)
print(c)
```

```python
# important function of tuple
#1.len()
t=(1,2,3,4,8,7,5)
print(len(t))
```

```python
#count()
t=(1,2,1,2,5,1,3,5,7,8)
print(t.count(1))
print(t.count(5))
```

```python
#3.index()
t=(10,20,10,30,10,40)
print(t.index(10))
print(t.index(30))
# print(t.index(50))
print(t.index(10,1,5))
```

```python
#4.sorted()
print(sorted(t))
print(tuple(sorted(t)))
print(sorted(t,reverse=True))
print(tuple(reversed(t)))  #reversed
```
